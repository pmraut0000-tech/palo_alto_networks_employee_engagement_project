# Employee Engagement, Satisfaction, and Burnout Diagnostic Analysis
### Illustrative case project: Palo Alto Networks

**Important:** This notebook uses synthetic data generated for academic/portfolio purposes. It is not based on confidential or actual Palo Alto Networks employee records.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DATA = Path('../data/employee_engagement_synthetic.csv')
df = pd.read_csv(DATA)
df.head()
```

```python
df[['Engagement_Score','Satisfaction_Score','Burnout_Score','Intent_to_Stay']].describe().round(2)
```

```python
dept = df.groupby('Department')[['Engagement_Score','Satisfaction_Score','Burnout_Score','Intent_to_Stay']].mean().round(2)
dept.sort_values('Burnout_Score', ascending=False)
```

```python
df['Burnout_Risk'].value_counts().plot(kind='bar', title='Burnout Risk Distribution')
plt.ylabel('Employees')
plt.show()
```

```python
cols = ['Engagement_Score','Satisfaction_Score','Burnout_Score','Intent_to_Stay','Manager_Support','Recognition','Career_Growth','Work_Life_Balance','Psychological_Safety','Compensation_Satisfaction','Weekly_Hours']
df[cols].corr().round(2)
```

## Interpretation Framework
- High engagement + high satisfaction + low burnout = healthy employee experience.
- High burnout with long working hours may indicate workload or capacity pressure.
- Low career-growth or recognition scores can weaken engagement and retention intent.
- Manager support and psychological safety are useful diagnostic dimensions for targeted action.
